program to withdraw money from atm

```python
pin=2260
avl_bal=20000
withdraw=int(input("enter the amount to be withdrawn:"))

if withdraw > avl_bal:
  print("sorry! not sufficient balance :(")
else:
  attempts = 3  # Allow 3 attempts for PIN entry
  for i in range(attempts):
    pin_ent = int(input("please enter your pin:"))
    if pin_ent == pin:
      print("please collect your cash")
      break
    else:
      print(f"incorrect pin, please try again:({attempts - 1 - i} attempts remaining)")"""where i is the index of the current attempt(eg-)
      first wrong attempt(i=0)
      so 3-1-0=2"""
  else: # This else block executes if the loop completes without a break
    print("Too many incorrect PIN attempts. Transaction cancelled.")
```

```python
"""program to print grades of the students ,
 A if 90-100
 B if 80-90
 C if 70-80
 D if 60-70
 E if 50-60
 f if failed"""

marks=int(input(" enter the total marks obtained by student :"))
total=int(input("enter the total marks :"))
percentage=marks/total*100
print("total percentage scored by the student is:",
      percentage)
if percentage>=90:
   print("grade A")
elif percentage>=80:
   print("grade B")
elif percentage>=70:
   print("grade C")
elif percentage>=60:
   print("grade D")
elif percentage>=50:
  print("grade E")
else:
   print("grade F - student failed")
```
